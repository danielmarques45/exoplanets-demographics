### Glossary
Hill radius the radial distance around a planet inside of which the planet’s own gravitational potential dominates the orbit of a test particle (compared to the gravitational potential of the star)

Kraft break the effective temperature that separates stars that do and do not harbour deep convective envelopes. Convection is needed to drive large scale magnetic dynamos, which in term affect stellar rotation such that stars hotter than the Kraft break at 6200 K exhibit rapid rotation velocities.

## 1. Abstract/Introduction 

Our capability of detecting exoplanets has significantly risen in the pasat decade, specially with the NASA's Kepler telescope operation, which led to a much deeper study of exoplanet demographics, allowing the discovery of several trends that are not observed at all in the Solar System. Acknowledging these patterns improves our comprehension of young stellar-planetary dynamics, and it will probably have an immense leap within the future NASA's Roman telescope's promising findings.


**Important notes:**
The clear division between exoplanets subpopulations is strongly tied to the inherent biases of the utilized detection methods, each of them being unable to cover a wide variety of exoplanetary system properties, invariably favouring particular stellar/planetary characteristics. 

The exoplanet population exhibits a much broader distribution of orbital eccentricities compared to the solar system’s planets, with multi-planet systems showing systematically lower eccentricity values than single planet systems, which is consistent with dynamical stability considerations.

• The exoplanet population exhibits a much broader distribution of orbital obliquities compared to the solar system’s planets, including numerous examples of hot giant planets with nearly polar orbits as well as retrograde orbits.

• Planets’ sizes and orbits are expected to evolve over time with particularly rapid evolution taking place during the earliest stages of the system’s lifetime. However, most stars are highly active at young ages, which stymies exoplanet detections and makes comparisons between young to old planetary systems difficult.

Kepler space telescope's goal was to detect Earth-sized planets in the habitable zones of Sun-like stars. The telescope wais mainly designed for the transit detection method (brightness curves) and has profoundly transformed the exoplanets demographics field due to its 3000+ confirmed exoplanets discovery.

## 2. Population-level trends in planets’ physical parameters

### 2.1 Radius valley

An interesting fact observed in close-in planets (short periods planets) is the clear distinction of two sub-populations separated by the lack of planets with $1.7-1.9R_{⊕}$. This dearth is known as the Radius Valley, marking the distinction between terrestrial super-Earths and larger sub-Neptunes whose bulk compositions are rich in volatiles (e.g. H-He, water, etc.) Undersanding the bulk composition of a planet might hint its origins, possibly leading to a further comprehension of its system's primordial dust cloud, and also contributes to investigating the possibility of an active dynamo, crucial to protect the planet's atmosphere and, consequently, to permit the emergence and sustenance of life.
