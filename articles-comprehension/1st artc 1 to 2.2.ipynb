## 1. Abstract/Introduction 

Our capability of detecting exoplanets has significantly risen in the pasat decade, specially with the NASA's Kepler telescope operation, which led to a much deeper study of exoplanet demographics, allowing the discovery of several trends that are not observed at all in the Solar System. Acknowledging these patterns improves our comprehension of young stellar-planetary dynamics, and it will probably have an immense leap within the future NASA's Roman telescope's promising findings.


**Important notes:**
The clear division between exoplanets subpopulations is strongly tied to the inherent biases of the utilized detection methods, each of them being unable to cover a wide variety of exoplanetary system properties, invariably favouring particular stellar/planetary characteristics. 

The exoplanet population exhibits a much broader distribution of orbital eccentricities compared to the solar system’s planets, with multi-planet systems showing systematically lower eccentricity values than single planet systems, which is consistent with dynamical stability considerations.

• The exoplanet population exhibits a much broader distribution of orbital obliquities compared to the solar system’s planets, including numerous examples of hot giant planets with nearly polar orbits as well as retrograde orbits.

• Planets’ sizes and orbits are expected to evolve over time with particularly rapid evolution taking place during the earliest stages of the system’s lifetime. However, most stars are highly active at young ages, which stymies exoplanet detections and makes comparisons between young to old planetary systems difficult.

Kepler space telescope's goal was to detect Earth-sized planets in the habitable zones of Sun-like stars. The telescope wais mainly designed for the transit detection method (brightness curves) and has profoundly transformed the exoplanets demographics field due to its 3000+ confirmed exoplanets discovery.

## 2. Population-level trends in planets’ physical parameters

### 2.1 Radius valley

An interesting fact observed in close-in planets (short periods planets) is the clear distinction of two sub-populations separated by the lack of known planets with $1.7-1.9R_{⊕}$. This dearth is known as the Radius Valley, marking the distinction between terrestrial super-Earths and larger sub-Neptunes whose bulk compositions are rich in volatiles (e.g. H-He, water, etc.) Although these studies are extremely difficult for sub-Neptunians due to interior degeneracies, undersanding the bulk composition of a planet might hint its origins, possibly leading to a further comprehension of its system's primordial dust cloud, and also contributes to investigating the possibility of an active dynamo, crucial to protect the planet's atmosphere and, consequently, to permit the emergence and sustenance of life. 

The origin of the Radius Valley is still an open debate. One of the possible explanations is that the valley is a “predetermined”  characteristic of super-Earths and Sub-Neptunes planetary systems: they form as two distinct subpopulations, at different timescales, whose superposition causes the dearth observed in their radii. Super-Earths tend to form in a gas-depleted disk, likely after the formation of another gas-rich planet; however, gas-poor super-Earths may form simultaneously with the gas-enveloped sub-Neptunes but with H-He accretion that is limited by the maximum isothermal envelope that can be accreted onto low-mass planetary cores ($≲ 1 − 2 M⊕$). This limitation and the dynamic interaction between these two bodies may induce a “prohibited zone" of planetary radius. 

This explanation presumes that both super-Earths and sub-Neptunes have terrestrial cores, but its also possible that sub-Neptunes are water worlds instead, formed beyond the ice line and at a later stage migrated to a close-in orbit. This composition affects the H-He accretion limit, likely restricting the planet radius.

### Mass-radius and mass-density relations

Measuring several planets' radii $R_p$ and masses $M_p$ by transit and other detection  methods allows us to compute mass-radius $M_p - R_p$ and mass-density $M_p - \rho_p$ relations (in fact, average-density of an assumed spherical bulk). Cataloging these data allows for more reliable predictions of masses from discovered planets whose masses are unknown and of bulk compositions (rerferring back to the debate on viability of life, existemnce of an active dynamo etc).

Aprropriate fits and precise functional forms of $R_p(M_p)$ and $\rho_p(M_p)$ have been vastly studied over the years, resulting in signifcantly different expressions due to the several possible different approaches: the sample of systems studied (including the Solar System or not), fitting techniques, depth of the variables taken in account (temperature of the planet, whether its a single or multi-planet system), etc. Thus, comparing these distinct empirical relations would surely be imprecise, what motivates a broader view of the relations.

As observed in the following graph, both $M_p - R_p$ and $M_p - \rho_p$ relations are fit by a broken power-law, with the first break corresponding to the transition from terrestrial objects to volatile-rich sub-Neptunes ($∼ 4 M_⊕$), being $~1,8_⊕$ the threshold to terrestrial planets, coinciding with the Radius Valley (reinforcing the idea of super-Earths and sub-Neptunes as two distinct subpopulations). The obtained scale is $R_p∝ M_{p}^{0.27}$, shallower than the expected $R_p∝ M_{p}^{\frac{1}{3}}$ for constant density. It turns out that terrestrial planets do not have constant densities, because at extreme pressures, the compounds in the planet's interior are compressed, leading to greater densities as the planet's mass (and thus its internal pressure) increases, exactly as observed in the lower panel of the graph.  

The second break at the transition between Neptunian worlds and Jovian worlds ($∼ 100 − 200 M_⊕$. As discussed before, the bulk of Neptunian worlds is unknown, despite the certainty of the abundance of volatile compounds. As these planets accrete votalite-rich material, their masses increase and their radii too, but way more rapidly than terrestrial planets. H-He envelopes grow the planets size in an immense rate, far greater than the mass accretion rate, justifying the positive relation $R_p ∝ M_{p}^{0.67}$ and the negative relation $M_p - \rho_p$. It is also important to note that volatile envelopes are much more sensitive to temperature than terrestrial worlds, what explains the dispersion of the data. 

A material's equation of state (EOS) describes how its density changes in pressure and temperature (i.e $ρ(P, T )$) At large masses $≳ 180 M⊕$ – characteristic of Jovian and massive Neptunians worlds, which have correspondingly high internal pressures – the EOS of cold H-He causes massive gas giant planets to become severely compressed. leading to a rapid density increase as the mass increases. Conversely, mass-radius has a slightly negative scaling $R_p ∝ M_{P}^{−0.06}$ due to the compressibility of H-He envelopes, making Jovian planets exhibit comparable sizes over nearly two orders of magnitude in mass (from $∼ 159 M_⊕$ to $∼ 4000 M_⊕$).
